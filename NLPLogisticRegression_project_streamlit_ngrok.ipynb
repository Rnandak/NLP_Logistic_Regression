```python
lemmatize = WordNetLemmatizer()
```

```python
def lemmatize_tokens(tokens):
    lemma =[]

    for token in tokens:
        lemma.append(lemmatize.lemmatize(token))
    return lemma
```

```python
NLPDS['lemmatize_summary'] = NLPDS['summary_tokenised'].apply(lemmatize_tokens)
```

```python
NLPDS['lemmatize_title'] = NLPDS['title_tokenised'].apply(lemmatize_tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
def identity_tokenizer(text):
    return text
```

```python
LRNLP = Pipeline([
    ('tdf',TfidfVectorizer(tokenizer=identity_tokenizer,preprocessor=identity_tokenizer)),
    ('LR', LogisticRegression(max_iter=1000))
])
```

```python
outcome = NLPDS['source_condition_query']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(feature,outcome,test_size=0.2,random_state=42)
```

```python
import re
import gradio as gr

def predict_disease(summary_text):
    if not summary_text.strip():
        return "Please enter a valid clinical trial summary."

    # 1. Lowercase
    cleaned = summary_text.lower()
    # 2. Clean punctuation
    cleaned = re.sub(r'[^a-zA-Z0-9\s]', '', cleaned)
    # 3. Tokenize
    tokens = word_tokenize(cleaned)
    # 4. Lemmatize
    lemmatized = lemmatize_tokens(tokens)

    # The pipeline expects a list/series of token lists as we used an identity tokenizer
    prediction = LRNLP.predict([lemmatized])[0]
    return prediction

# Create and launch the Gradio interface
demo = gr.Interface(
    fn=predict_disease,
    inputs=gr.Textbox(lines=5, label="Clinical Trial Brief Summary", placeholder="Enter trial brief summary here..."),
    outputs=gr.Textbox(label="Predicted Disease / Condition Query"),
    title="Clinical Trials Disease Predictor",
    description="Enter a clinical trial summary to predict its target disease or condition category."
)

demo.launch(share=True)
```

```python
%%writefile app.py
import streamlit as st
import joblib
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Ensure NLTK resources are available
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')

# Initialize Lemmatizer
lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(token) for token in tokens]

# Define the identity tokenizer used in the pipeline
def identity_tokenizer(text):
    return text

# Page Configuration
st.set_page_config(
    page_title="Clinical Trials Disease Predictor",
    page_icon="⚕ ",
    layout="centered"
)

# Load the model
@st.cache_resource
def load_model():
    # Make sure 'identity_tokenizer' is in the global scope so joblib can deserialize it
    return joblib.load('clinical_trials_model.joblib')

try:
    model = load_model()
    model_loaded = True
except Exception as e:
    model_loaded = False
    st.error(f"Error loading model 'clinical_trials_model.joblib': {e}")
    st.info("Please ensure the trained 'clinical_trials_model.joblib' file is in the same directory as this script.")

# App UI
st.title("⚕  Clinical Trials Disease Predictor")
st.write("Enter a clinical trial summary below to predict its target disease or condition category.")

summary_text = st.text_area(
    label="Clinical Trial Brief Summary",
    placeholder="Enter trial brief summary here...",
    height=200
)

if st.button("Predict Condition", type="primary"):
    if not summary_text.strip():
        st.warning("Please enter a valid clinical trial summary.")
    elif not model_loaded:
        st.error("Model is not loaded. Cannot perform prediction.")
    else:
        with st.spinner("Processing text and predicting..."):
            # 1. Lowercase
            cleaned = summary_text.lower()

            # 2. Clean punctuation
            cleaned = re.sub(r'[^a-zA-Z0-9\s]', '', cleaned)

            # 3. Tokenize
            tokens = word_tokenize(cleaned)

            # 4. Lemmatize
            lemmatized = lemmatize_tokens(tokens)

            # 5. Predict using loaded pipeline
            prediction = model.predict([lemmatized])[0]

            # Display Result
            st.success(f"**Predicted Disease / Condition Query:** {prediction}")
```

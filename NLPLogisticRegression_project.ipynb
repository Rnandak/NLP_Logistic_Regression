```python
from nltk.tokenize import word_tokenize
```

```python
NLPDS['summary_tokenised']=NLPDS['brief_summary_clean'].apply(word_tokenize)
```

```python
NLPDS['title_tokenised'] = NLPDS['title'].apply(word_tokenize)
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatize = WordNetLemmatizer()
```

```python
def lemmatize_tokens(tokens):
    lemma =[]

    for token in tokens:
        lemma.append(lemmatize.lemmatize(token))
    return lemma
```

```python
NLPDS['lemmatize_summary'] = NLPDS['summary_tokenised'].apply(lemmatize_tokens)
```

```python
NLPDS['lemmatize_title'] = NLPDS['title_tokenised'].apply(lemmatize_tokens)
```

```python
NLPDS
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
def identity_tokenizer(text):
    return text
```

```python
LRNLP = Pipeline([
    ('tdf',TfidfVectorizer(tokenizer=identity_tokenizer,preprocessor=identity_tokenizer)),
    ('LR', LogisticRegression(max_iter=1000))
])
```

```python
feature = NLPDS['lemmatize_summary']
```

```python
outcome = NLPDS['source_condition_query']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(feature,outcome,test_size=0.2,random_state=42)
```

```python
print(type(xtrain))
print(type(ytrain))

print(len(xtrain))
print(len(ytrain))
```

```python
print(xtrain.shape)
```

```python
print(xtrain.head())
```

```python
LRNLP.fit(xtrain,ytrain)
```

```python
ypred = LRNLP.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
display(accuracy_score(ytest,ypred))
```

```python
display(classification_report(ytest,ypred))
```

```python
display(confusion_matrix(ytest,ypred))
```

```python
import re
import gradio as gr

def predict_disease(summary_text):
    if not summary_text.strip():
        return "Please enter a valid clinical trial summary."

    # 1. Lowercase
    cleaned = summary_text.lower()
    # 2. Clean punctuation
    cleaned = re.sub(r'[^a-zA-Z0-9\s]', '', cleaned)
    # 3. Tokenize
    tokens = word_tokenize(cleaned)
    # 4. Lemmatize
    lemmatized = lemmatize_tokens(tokens)

    # The pipeline expects a list/series of token lists as we used an identity tokenizer
    prediction = LRNLP.predict([lemmatized])[0]
    return prediction

# Create and launch the Gradio interface
demo = gr.Interface(
    fn=predict_disease,
    inputs=gr.Textbox(lines=5, label="Clinical Trial Brief Summary", placeholder="Enter trial brief summary here..."),
    outputs=gr.Textbox(label="Predicted Disease / Condition Query"),
    title="Clinical Trials Disease Predictor",
    description="Enter a clinical trial summary to predict its target disease or condition category."
)

demo.launch(share=True)
```

```python
import torch
```

```python
X = torch.tensor([[5.1, 3.5, 1.4, 0.2]], dtype=torch.float32)
```

```python
y=torch.tensor([[1.0]])
```

```python
y
```

```python
W = torch.randn((4,1), requires_grad=True)
b = torch.randn((1), requires_grad=True)
```

```python
b
```

```python
z = X @ W + b
```

```python
output = torch.sigmoid(z)
```

```python
print("Prediction:", output.item())
```

```python
loss = torch.mean((y - output)**2)
```

```python
print("Loss:", loss.item())
```

```python
loss.backward()
```

```python
ds = torch.tensor([
    [2, 6, 5, 7],   # Candidate 1
    [8, 9, 8, 8],   # Candidate 2
    [1, 4, 3, 5],   # Candidate 3
    [5, 8, 9, 6]    # Candidate 4
], dtype=torch.float32)
```

```python
y = torch.tensor([
    [0],
    [1],
    [0],
    [1]
], dtype=torch.float32)
```

```python

weights = torch.randn((4,1), requires_grad=True)

bias = torch.randn(1, requires_grad=True)
```

```python
print("Initial Weights")
print(weights)
```

```python
z = X @ weights + bias
```

```python
predictions = torch.sigmoid(z)

print("\nPredictions")
print(predictions)
```

```python
loss = torch.mean(
    (y - predictions) ** 2
)
```

```python
print("\nLoss")
print(loss.item())
```

```python
loss.backward()

print("\nGradient For Each Weight")
print(weights.grad)

print("\nBias Gradient")
print(bias.grad)
```
